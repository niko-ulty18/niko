```python
#def
import random
import time
import sys

sys.setrecursionlimit(20000)

# --- ESTRUCTURA DE LA LISTA ENLAZADA ---
class Nodo:
    def __init__(self, dato):
        self.dato = dato
        self.siguiente = None

class ListaEnlazada:
    def __init__(self):
        self.cabeza = None
        self.esta_ordenada = False

    def insertar_final(self, dato):
        nuevo = Nodo(dato)
        self.esta_ordenada = False
        if not self.cabeza:
            self.cabeza = nuevo
            return
        actual = self.cabeza
        while actual.siguiente:
            actual = actual.siguiente
        actual.siguiente = nuevo

    def eliminar_lista(self):
        self.cabeza = None
        self.esta_ordenada = False

    def mostrar(self, resaltado=None):
        if not self.cabeza:
            print("\n[!] La lista está vacía.")
            return
        actual = self.cabeza
        elementos = []
        while actual:
            val = actual.dato
            # Si el valor coincide con el pivote resaltado, lo marcamos visualmente
            if resaltado is not None and val == resaltado:
                elementos.append(f"[{val}]*")
            else:
                elementos.append(str(val))
            actual = actual.siguiente

        if len(elementos) > 20:
            print(" -> ".join(elementos[:20]) + " ...")
        else:
            print(" -> ".join(elementos))

    def a_lista_python(self):
        res = []
        act = self.cabeza
        while act:
            res.append(act.dato)
            act = act.siguiente
        return res

    def reconstruir(self, lista_py):
        self.cabeza = None
        if not lista_py: return
        self.cabeza = Nodo(lista_py[0])
        actual = self.cabeza
        for d in lista_py[1:]:
            actual.siguiente = Nodo(d)
            actual = actual.siguiente
        self.esta_ordenada = True

    def eliminar_lista(self):
        self.cabeza = None
        self.esta_ordenada = False



# --- ALGORITMOS QUICKSORT CON IDENTIFICACIÓN DE PIVOTE ---

def obtener_pivote(arr, tipo):
    if not arr: return None
    if tipo == "1": # Inicial
        return arr[0]
    elif tipo == "2": # Aleatorio
        return random.choice(arr)
    elif tipo == "3": # Mediana
        primero = arr[0]
        medio = arr[len(arr)//2]
        ultimo = arr[-1]
        return sorted([primero, medio, ultimo])[1]
    return arr[0]

def quicksort(arr, tipo_pivote):
    if len(arr) <= 1: return arr

    # En cada recursión se elige el pivote según el método
    if tipo_pivote == "1": p = arr[0]
    elif tipo_pivote == "2": p = random.choice(arr)
    else:
        primero = arr[0]
        medio = arr[len(arr)//2]
        ultimo = arr[-1]
        p = sorted([primero, medio, ultimo])[1]

    izq = [x for x in arr if x < p]
    centro = [x for x in arr if x == p]
    der = [x for x in arr if x > p]

    return quicksort(izq, tipo_pivote) + centro + quicksort(der, tipo_pivote)

# --- MENU ---
mi_lista = ListaEnlazada()
mantener_ordenada = False
historial_tiempos = {"Inicial": None, "Aleatorio": None, "Mediana": None}

def menu():
    global mantener_ordenada
    while True:
        print("\n======= GESTIÓN DE LISTA =======")
        print(f" Estado: {'✅ Ordenada' if mi_lista.esta_ordenada else '❌ Desordenada'}")
        print(f" Modo: {'💾 GUARDAR' if mantener_ordenada else '🔄 VISTA PREVIA (No guarda)'}")
        print(" 1. Generar lista aleatoria")
        print(" 2. Ver la lista")
        print(" 3. Seleccionar Pivote y Ordenar")
        print(" 4. Vaciar lista")
        print(" 5. Comparar tiempos")
        print(" 6. Cambiar modo (Guardar/Vista)")
        print(" 7. Salir")

        op = input("\nSeleccione opción: ")

        if op == "1":
            try:
                tam = int(input("Tamaño: "))
                mi_lista.eliminar_lista()
                for _ in range(tam):
                    mi_lista.insertar_final(random.randint(1, 100))
                print(f"✅ Generados {tam} números.")
            except: print("❌ Error.")

        elif op == "2":
            mi_lista.mostrar()

        elif op == "3":
            datos_py = mi_lista.a_lista_python()
            if not datos_py:
                print("⚠️ Lista vacía."); continue

            print("\n--- SELECCIÓN DE PIVOTE ---")
            print("1. Inicial ")
            print("2. Aleatorio ")
            print("3. Mediana ")
            p_op = input("Seleccione tipo de pivote: ")

            # --- VISTA PREVIA DEL PIVOTE ---
            pivote_elegido = obtener_pivote(datos_py, p_op)
            nombre_metodo = {"1": "Inicial", "2": "Aleatorio", "3": "Mediana"}.get(p_op, "Aleatorio")

            print(f"\n🔍 VISTA PREVIA DEL PIVOTE ({nombre_metodo}):")
            print(f"Pivote seleccionado: {pivote_elegido}")
            print("Lista con pivote resaltado:")
            mi_lista.mostrar(resaltado=pivote_elegido)

            input("\nPresione Enter para comenzar el ordenamiento...")

            # --- ORDENAMIENTO ---
            inicio = time.time()
            ordenados = quicksort(datos_py, p_op)
            fin = time.time()

            tiempo = fin - inicio
            historial_tiempos[nombre_metodo] = tiempo

            if mantener_ordenada:
                mi_lista.reconstruir(ordenados)
                print("✅ Cambios guardados en la lista.")
            else:
                print("✅ Ordenamiento completado (solo vista previa).")

            print(f"⏱ Tiempo: {tiempo:.6f} seg.")

        elif op == "4":
            mi_lista.eliminar_lista()
            print("\n🗑️  La lista ha sido vaciada.")

        elif op == "5":
            print("\n--- COMPARATIVA ---")
            for m, t in historial_tiempos.items():
                if t: print(f"• {m}: {t:.6f}s")

        elif op == "6":
            mantener_ordenada = not mantener_ordenada
            print(f"Modo cambiado.")

        elif op == "7": break

if __name__ == "__main__":
    menu()
```
